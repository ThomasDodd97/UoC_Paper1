```python
Values = np.linspace(np.min(np.array(df["DeltaPolymerMass_pct"])),np.max(np.array(df["DeltaPolymerMass_pct"])),5)
Values
```

```python
Indices = []
for i in Values:
    Closeness = []
    for j in np.array(df["DeltaPolymerMass_pct"]):
        Closeness.append(np.abs(j-i))
    Index = np.argmin(Closeness)
    Indices.append(Index)
Indices
```
