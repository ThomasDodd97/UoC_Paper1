```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
sample = np.array([2,3,4])
PreCureMass = np.array([3.41897,3.21978,3.4586])
PostCureMass = np.array([3.15118,2.93008,3.15616])
CureMassLoss = PreCureMass - PostCureMass
CureMassLossPct = (CureMassLoss/PreCureMass)*100
PostThfMass = [0.90733,2.89547,2.64668]
ThfInsolubleFraction = (PostThfMass/PostCureMass)*100
CureMassLossPct
```

```python
plt.scatter(CureMassLossPct,ThfInsolubleFraction,label="Samples",color="black")
plt.xlabel(r"Cured Mass Loss, $\Delta m_{\%}$ (%)",fontsize="14")
plt.ylabel("Gel Fraction (%)",fontsize="14")
params = np.polyfit(x=CureMassLossPct,y=ThfInsolubleFraction,deg=1)
print(params)
def f(x,m,c):
    return m*x+c
test_x = np.linspace(np.min(CureMassLossPct),np.max(CureMassLossPct),10)
test_y = []
for x in test_x:
    test_y.append(f(x,params[0],params[1]))
plt.plot(test_x,test_y,linestyle="dotted",label=r"Linear Fit",color="#0077b6")
plt.ylim(0,100)
plt.yticks(np.linspace(0,100,11),fontsize="14")
plt.yticks(np.linspace(0,100,21),minor=True,fontsize="14")
plt.xlim(7,10)
plt.xticks(np.linspace(7,10,7),fontsize="14")
plt.xticks(np.linspace(7,10,13),minor=True,fontsize="14")
plt.text(x=7.85,y=40,s=r"$y=60.2x-442.5$",rotation=53.8,color="#0077b6",fontsize="14")
plt.text(x=7.8,y=22,s="2",color="black",fontsize="14")
plt.text(x=8.71,y=77,s="4",color="black",fontsize="14")
plt.text(x=8.97,y=92,s="3",color="black",fontsize="14")
plt.legend(fontsize="14",loc="upper left")
plt.tight_layout()
plt.savefig(fname='MassLossInvestigation.png',dpi=1000,format='png')
```
