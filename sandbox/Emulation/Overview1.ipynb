```python
GpMa05 = [Pr_GpMa05,Qr_GpMa05,Se_GpMa05_Mipt,Se_GpMa05_BayPi,Se_GpMa05_BayEi]
GpMa15 = [Pr_GpMa15,Qr_GpMa15,Se_GpMa15_Mipt,Se_GpMa15_BayPi,Se_GpMa15_BayEi]
GpMa25 = [Pr_GpMa25,Qr_GpMa25,Se_GpMa25_Mipt,Se_GpMa25_BayPi,Se_GpMa25_BayEi]
GpRbf = [Pr_GpRbf,Qr_GpRbf,Se_GpRbf_Mipt,Se_GpRbf_BayPi,Se_GpRbf_BayEi]
Xgb = [Pr_Xgb,Qr_Xgb,Se_Xgb_Mipt,Se_Xgb_BayPi,Se_Xgb_BayEi]

Datasets = [GpMa05,GpMa15,GpMa25,GpRbf,Xgb]
```

```python
DatasetsMeans = []
for dataset in Datasets:
    DatasubsetMeans = []
    for datasubset in dataset:
        DatasubsetMeans.append(np.mean(datasubset))
    DatasetsMeans.append(DatasubsetMeans)
DatasetsMeans = np.array(DatasetsMeans)
```

```python
DataPoints = []
EmulatorTypes = []
OptimisationTypes = []
for FirstCounter,dataset in enumerate(Datasets):
    # e.g. dataset = GpMa05
    if FirstCounter == 0:
        EmulatorType = "Gp_Ma05"
    elif FirstCounter == 1:
        EmulatorType = "Gp_Ma15"
    elif FirstCounter == 2:
        EmulatorType = "Gp_Ma25"
    elif FirstCounter == 3:
        EmulatorType = "Gp_Rbf"
    elif FirstCounter == 4:
        EmulatorType = "Xgb"
    for SecondCounter,datasubset in enumerate(dataset):
        # e.g. datasubset = Pr_GpMa05
        if SecondCounter == 0:
            OptimisationType = "Pseudorandom"
        elif SecondCounter == 1:
            OptimisationType = "Quasirandom"
        elif SecondCounter == 2:
            OptimisationType = "Sequential MIPT"
        elif SecondCounter == 3:
            OptimisationType = "Sequential PI"
        elif SecondCounter == 4:
            OptimisationType = "Sequential EI"
        for DataPoint in datasubset:
            DataPoints.append(DataPoint)
            EmulatorTypes.append(EmulatorType)
            OptimisationTypes.append(OptimisationType)
```

```python
d = {"DataPoints":DataPoints,"EmulatorTypes":EmulatorTypes,"OptimisationTypes":OptimisationTypes}
df = pd.DataFrame(data=d)
df
```

```python
SomeNumbers = []
for i in DatasetsMeans:
    for j in i:
        SomeNumbers.append(j)
OtherNumbers = [0,1,2,3,4]
SomeMoreNumbers = [-0.32,-0.16,0,0.16,0.32]
LastOfTheNumbers = []
for i in OtherNumbers:
    for j in SomeMoreNumbers:
        LastOfTheNumbers.append(i+j)
```

```python
plt.gcf().set_size_inches(10.25, 4)
ax = sns.boxplot(data=df,x="EmulatorTypes", y="DataPoints", hue="OptimisationTypes")
sns.move_legend(ax,loc="upper left", bbox_to_anchor=(-0.05, 1.2),ncols=5)
ax.scatter(x=LastOfTheNumbers,y=SomeNumbers,marker="D",s=15,color="white",zorder=4,label="Mean",edgecolors="black")
ax.scatter(x=[0],y=[0],marker=".",s=130,color="white",edgecolor="black",label="Outlier")
ax.set_xticks(range(5))
ax.set_xticklabels([r"$\mathcal{GP}$"+"\nMatérn Kernel\n"+r"$\nu=0.5$",
                    r"$\mathcal{GP}$"+"\nMatérn Kernel\n"+r"$\nu=1.5$",
                    r"$\mathcal{GP}$"+"\nMatérn Kernel\n"+r"$\nu=2.5$",
                    r"$\mathcal{GP}$"+"\n Radial Basis Function",
                    "GBDT"])
ax.set_xlabel("Emulator Type",fontsize=13)
ax.set_ylabel(r'Best Percentage Mass Loss, $\Delta$m$^\circ_\%$ (%)      ',fontsize=13)
ax.legend(bbox_to_anchor=(1.01, 1.2),ncols=7)
ax.set_ylim(10,20)
plt.tight_layout()
plt.savefig(fname='ComputationalResultOverview.png',dpi=1000,format='png')
```
