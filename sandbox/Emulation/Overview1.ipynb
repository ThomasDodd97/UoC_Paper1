```python
GpMa05 = [Pr_GpMa05,Qr_GpMa05,Se_GpMa05_Mipt,Se_GpMa05_BayPi,Se_GpMa05_BayEi]
GpMa15 = [Pr_GpMa15,Qr_GpMa15,Se_GpMa15_Mipt,Se_GpMa15_BayPi,Se_GpMa15_BayEi]
GpMa25 = [Pr_GpMa25,Qr_GpMa25,Se_GpMa25_Mipt,Se_GpMa25_BayPi,Se_GpMa25_BayEi]
GpRbf = [Pr_GpRbf,Qr_GpRbf,Se_GpRbf_Mipt,Se_GpRbf_BayPi,Se_GpRbf_BayEi]
Xgb = [Pr_Xgb,Qr_Xgb,Se_Xgb_Mipt,Se_Xgb_BayPi,Se_Xgb_BayEi]

Datasets = [GpMa05,GpMa15,GpMa25,GpRbf,Xgb]
```

```python
DatasetsMeans = []
for dataset in Datasets:
    DatasubsetMeans = []
    for datasubset in dataset:
        DatasubsetMeans.append(np.mean(datasubset))
    DatasetsMeans.append(DatasubsetMeans)
DatasetsMeans = np.array(DatasetsMeans)
```

```python
DatasetsStds = []
for dataset in Datasets:
    DatasubsetStds = []
    for datasubset in dataset:
        DatasubsetStds.append(np.std(datasubset))
    DatasetsStds.append(DatasubsetStds)
DatasetsStds = np.array(DatasetsStds)
```

```python
Emulators_Types = ["Gp_Ma05","Gp_Ma15","Gp_Ma25","Gp_Rbf","Xgb"]
Samplers_Types = ["Pseudorandom","Quasirandom","Sequential MIPT","Sequential PI","Sequential EI"]
```

```python
DataPoints = []
EmulatorTypes = []
OptimisationTypes = []
Hues = []
for FirstCounter,dataset in enumerate(Datasets):
    # e.g. dataset = GpMa05
    if FirstCounter == 0:
        EmulatorType = "Gp_Ma05"
        # Hue = ColorScheme[0]
    elif FirstCounter == 1:
        EmulatorType = "Gp_Ma15"
        # Hue = ColorScheme[1]
    elif FirstCounter == 2:
        EmulatorType = "Gp_Ma25"
        # Hue = ColorScheme[2]
    elif FirstCounter == 3:
        EmulatorType = "Gp_Rbf"
        # Hue = ColorScheme[3]
    elif FirstCounter == 4:
        EmulatorType = "Xgb"
        # Hue = ColorScheme[4]
    for SecondCounter,datasubset in enumerate(dataset):
        # e.g. datasubset = Pr_GpMa05
        if SecondCounter == 0:
            OptimisationType = "Pseudorandom"
        elif SecondCounter == 1:
            OptimisationType = "Quasirandom"
        elif SecondCounter == 2:
            OptimisationType = "Sequential MIPT"
        elif SecondCounter == 3:
            OptimisationType = "Sequential PI"
        elif SecondCounter == 4:
            OptimisationType = "Sequential EI"
        for DataPoint in datasubset:
            DataPoints.append(DataPoint)
            EmulatorTypes.append(EmulatorType)
            OptimisationTypes.append(OptimisationType)
            # Hues.append(Hue)
```

```python
d = {"DataPoints":DataPoints,"EmulatorTypes":EmulatorTypes,"OptimisationTypes":OptimisationTypes}
df = pd.DataFrame(data=d)
df
```

```python
Max_lis = [18.57,18.27,18.29,15.14,19.14]
Min_lis = [10.7,10.77,10.8,10.81,9.89]
MaxMinus10_lis = []
for maxi,mini in zip(Max_lis,Min_lis):
    Diff = maxi-mini
    MaxMinus10_lis.append(mini+(Diff*0.90))
MaxMinus10_lis
```

```python
ColorScheme = ["#ee9b00","#ca6702","#94d2bd","#0a9396","#005f73"]
ColorPalette = {"Pseudorandom": ColorScheme[0], "Quasirandom": ColorScheme[1],"Sequential MIPT":ColorScheme[2],"Sequential PI":ColorScheme[3],"Sequential EI":ColorScheme[4]}
```

```python
fig, ax = plt.subplots(1, 5)

ax[0].scatter(x=0,y=0,c=ColorScheme[0],s=60,label="PR: Pseudorandom",marker="s")
ax[0].scatter(x=0,y=0,c=ColorScheme[1],s=60,label="QR: Quasirandom",marker="s")
ax[0].scatter(x=0,y=0,c=ColorScheme[2],s=60,label="MIPT: mc-intersite-proj-th",marker="s")
ax[0].scatter(x=0,y=0,c=ColorScheme[3],s=60,label="PI: Probability of Improvement",marker="s")
ax[0].scatter(x=0,y=0,c=ColorScheme[4],s=60,label="EI: Expected Improvement",marker="s")
FirstLegend = ax[0].legend(ncols=6,bbox_to_anchor=(5.25,-0.2))

for i,emulator in enumerate(Emulators_Types):
    MaxLine = ax[i].axhline(y=Max_lis[i],xmin=0,xmax=1,linestyle="dashed",color="black",label="Global Maximum/Minimum Bound")
    MinLine = ax[i].axhline(y=Min_lis[i],xmin=0,xmax=1,linestyle="dashed",color="black")
    RegionLine = ax[i].axhline(y=MaxMinus10_lis[i],xmin=0,xmax=1,linestyle="dotted",color="black",label="Region of Optimality Lower Bound")
    for j,(OptTech,ArbXAxisValue) in enumerate(zip(OptimisationTypes,ArbXAxis)):
        if j == 1:
            Diamond = ax[i].scatter(ArbXAxisValue,DatasetsMeans[i][j],marker="D",s=25,color="white",edgecolors="black",label="Mean",zorder=5)
        else:
            ax[i].scatter(ArbXAxisValue,DatasetsMeans[i][j],marker="D",s=25,color="white",edgecolors="black",zorder=5)
    sns.stripplot(data=df.loc[(df['EmulatorTypes'] == Emulators_Types[i])],y="DataPoints", x="OptimisationTypes",hue="OptimisationTypes",ax=ax[i],s=0.5,jitter=0.4,palette=ColorPalette)
    ax[i].set_title(EmulatorNames[i])
    labels = OptimisationNames
    ticks = np.linspace(0,4,5)
    ax[i].set_xticks(ticks=ticks, labels=labels)
    ax[i].set(xlabel=None)
    if i == 0:
        ax[i].set_yticks([])
        ax[i].set_yticklabels([])
        ax[i].set(ylabel=None)
        ax[i].set_yticks(ticks=MajorTicks,minor=False)
        ax[i].set_yticks(ticks=MinorTicks,minor=True)
        ax[i].set_yticklabels(MajorTicks)
        ax[i].set_ylim(Lower,Upper)
    if i > 0:
        ax[i].set_yticks([])
        ax[i].set_yticklabels([])
        ax[i].set(ylabel=None)
        ax[i].set_yticks(ticks=MajorTicks,minor=False)
        ax[i].set_yticks(ticks=MinorTicks,minor=True)
        ax[i].set_ylim(Lower,Upper)

SecondLegend = ax[1].legend(handles=[Diamond,MaxLine,RegionLine],ncols=6,bbox_to_anchor=(2.05,-0.3))

fig.supxlabel(r"Sampling Technique",fontsize=15,x=0.5,y=-0.03)
fig.supylabel(r"Max Percentage Mass Loss, $\Delta$m$^\circ_\%$ (%)",fontsize=15,x=0.07,y=0.5)

plt.subplots_adjust(wspace=0.05)
plt.gcf().set_size_inches(12, 4)
# plt.tight_layout()
plt.savefig(fname='ComputationalResultOverview.png',dpi=1000,format='png',bbox_inches='tight')
```

```python
EmulatorSamplerPercentages = []
for EmulatorDataset,MaxMinus10 in zip(Datasets,MaxMinus10_lis):
    SamplerPercentages = []
    for SamplerDataset in EmulatorDataset:
        SamplesAboveBound = []
        for y_value in SamplerDataset:
            if y_value > MaxMinus10:
                SamplesAboveBound.append(y_value)
        SamplerPercentages.append(round((len(SamplesAboveBound)/len(SamplerDataset))*100,2))
    EmulatorSamplerPercentages.append(SamplerPercentages)
```
