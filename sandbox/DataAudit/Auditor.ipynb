```python
Consistent = []
for dataframe in dataframes:
    for count,(s1,s2,b1,ca,ia) in enumerate(zip(np.array(dataframe["s1"]),np.array(dataframe["s2"]),np.array(dataframe["b1"]),np.array(dataframe["CA_stoichiometry"]),np.array(dataframe["IA_stoichiometry"]))):
        ca_calc = (0.5+(2.0-0.5)*s1)*b1
        ia_calc = (1.0+(2.0-1.0)*s2)*(1-b1)
        ca_calc_round = round(ca_calc,1)
        ca_round = round(ca,1)
        ia_calc_round = round(ia_calc,1)
        ia_round = round(ia,1)
        if (ca_calc_round != ca_round) == True:
            Consistent.append("No")
            break
        elif (ia_calc_round != ia_round) == True:
            Consistent.append("No")
            break
        elif (count+1 == len(np.array(dataframe["s1"]))) == True:
            Consistent.append("Yes")
            break
```

```python
print("Name\tConsistent?")
for name,consistency in zip(dataframenames,Consistent):
    print(f"{name}:\t{consistency}")
```
