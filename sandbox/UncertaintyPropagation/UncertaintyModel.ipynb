# Uncertainty Progagation Calculations

```python
# Uncertainty Propagation for Mass Balance Measurements
import numpy as np

# This corresponds to one standard deviation of uncertainty around this mean.
MassBalanceRepeatability_g = 0.01

# Basic mass settings.
InitialMass_g_flt = 3
MinimumFinalMass_g_flt = 0.92*InitialMass_g_flt
MaximumFinalMass_g_flt = 0.79*InitialMass_g_flt

# Testing loop
StandardDeviationOfPercentageMassLosses = []

# Loop
tests = 10000
for i in range(tests):
    CurrentInitialMass_g_flt = InitialMass_g_flt
    CurrentFinalMass_g_flt = np.random.uniform(MinimumFinalMass_g_flt,MaximumFinalMass_g_flt)
    subtests = 10000
    CurrentMeasuredInitialMasses_g_arr = np.random.normal(loc=CurrentInitialMass_g_flt,scale=MassBalanceRepeatability_g,size=subtests)
    CurrentMeasuredFinalMasses_g_arr = np.random.normal(loc=CurrentFinalMass_g_flt,scale=MassBalanceRepeatability_g,size=subtests)
    CurrentMeasuredMassLoss_g_arr = CurrentMeasuredInitialMasses_g_arr-CurrentMeasuredFinalMasses_g_arr
    CurrentMeasuredPercentageMassLoss_g_arr = ((CurrentMeasuredInitialMasses_g_arr-CurrentMeasuredFinalMasses_g_arr)*100)/CurrentMeasuredInitialMasses_g_arr
    StandardDeviationOfPercentageMassLosses.append(np.std(CurrentMeasuredPercentageMassLoss_g_arr))

# Results
print("Expected Uncertainty (1σ):")
print(f"Mean = {round(np.average(StandardDeviationOfPercentageMassLosses),2)} %")
print(f"Standard Deviation (68%) = {round(np.std(StandardDeviationOfPercentageMassLosses),2)}")
print(f"Standard Deviation (95%) = {round(np.std(StandardDeviationOfPercentageMassLosses)*2,2)}")
```

```python
# An expected uncertainty can be ascertained for the one-shot techniques.
# Grid Maximum Values = 16.24
# Quasirandom Maximum = 14.55
# Pseudorandom Maximum = 14.01

InitialMass = 3
GridFinalMass = (1-0.1624)*InitialMass
QuasiFinalMass = (1-0.1455)*InitialMass
PseudoFinalMass = (1-0.1401)*InitialMass

tests = 1000000
CurrentMeasuredInitialMasses = np.random.normal(loc=InitialMass,scale=MassBalanceRepeatability_g,size=tests)
CurrentMeasuredFinalMasses = np.random.normal(loc=GridFinalMass,scale=MassBalanceRepeatability_g,size=tests)
CurrentMeasuredPercentageMassLoss_g_arr = ((CurrentMeasuredInitialMasses-CurrentMeasuredFinalMasses)*100)/CurrentMeasuredInitialMasses
GridAverage = np.average(CurrentMeasuredPercentageMassLoss_g_arr)
GridStandardDeviation = np.std(CurrentMeasuredPercentageMassLoss_g_arr)

CurrentMeasuredInitialMasses = np.random.normal(loc=InitialMass,scale=MassBalanceRepeatability_g,size=tests)
CurrentMeasuredFinalMasses = np.random.normal(loc=QuasiFinalMass,scale=MassBalanceRepeatability_g,size=tests)
CurrentMeasuredPercentageMassLoss_g_arr = ((CurrentMeasuredInitialMasses-CurrentMeasuredFinalMasses)*100)/CurrentMeasuredInitialMasses
QuasiAverage = np.average(CurrentMeasuredPercentageMassLoss_g_arr)
QuasiStandardDeviation = np.std(CurrentMeasuredPercentageMassLoss_g_arr)

CurrentMeasuredInitialMasses = np.random.normal(loc=InitialMass,scale=MassBalanceRepeatability_g,size=tests)
CurrentMeasuredFinalMasses = np.random.normal(loc=PseudoFinalMass,scale=MassBalanceRepeatability_g,size=tests)
CurrentMeasuredPercentageMassLoss_g_arr = ((CurrentMeasuredInitialMasses-CurrentMeasuredFinalMasses)*100)/CurrentMeasuredInitialMasses
PseudoAverage = np.average(CurrentMeasuredPercentageMassLoss_g_arr)
PseudoStandardDeviation = np.std(CurrentMeasuredPercentageMassLoss_g_arr)

print("Expected Uncertainty (1σ):")
print(f"Grid = {round(np.average(GridStandardDeviation),2)} %")
print(f"Quasirandom = {round(np.average(QuasiStandardDeviation),2)} %")
print(f"Pseudorandom = {round(np.average(PseudoStandardDeviation),2)} %")
```
